```python
import lmstudio as lms

# Conecta (padrão localhost:1234)
with lms.Client() as client:
    # Carrega/força com GPU offload (ratio = 1.0 ou "max" ≈ full offload)
    model = client.llm.load_new_instance(
        "oh-dcft-v3.1-gpt-4o-mini",
        config={
            "gpu": {
                "ratio": 1.0,          # 1.0 = máximo possível (full offload)
                # "ratio": 0.8,        # ou 80% das layers na GPU
                # "mainGpu": 0,        # se tiver múltiplas GPUs (opcional)
                # "tensorSplit": [1.0] # split se multi-GPU (opcional)
            },
            "contextLength": 8192,     # ajuste se quiser contexto maior
            # "offloadKVCacheToGpu": True,  # KV cache na GPU (acelera muito)
        }
    )

    # Agora gera com GPU
    response = model.respond("me da um codigo python de 40 linhas?")
    print(response.content)
```

```python
import lmstudio as lms

# Conecta no LM Studio (padrão: localhost:1234)
with lms.Client() as client:
    # Carrega o modelo com ~50% na GPU / 50% na CPU
    model = client.llm.load_new_instance(
        "oh-dcft-v3.1-gpt-4o-mini",   # ou o identificador exato que aparece no LM Studio
        config={
            "gpu": {
                "ratio": 0.5,              # 0.0 = tudo CPU   |  1.0 = máximo na GPU
                                           # 0.5  → metade-metade (o mais comum para balancear)
                # "mainGpu": 0,            # opcional: qual GPU principal (se tiver + de uma)
                # "tensorSplit": [0.5, 0.5]  # só se tiver 2+ GPUs e quiser dividir entre elas
            },
            "contextLength": 8192,         # ou o valor que você precisa
            # "offloadKVCacheToGpu": True, # opcional: deixa KV cache na GPU (acelera bastante)
                                           #               mas consome mais VRAM
        }
    )

    # Gera a resposta usando o modelo carregado
    response = model.respond("me da um codigo python de 40 linhas?")
    
    print("Resposta do modelo:")
    print(response.content)
    
    # Dica: depois que não precisar mais, pode descarregar
    # model.unload()   # libera memória (opcional)
```

```python
model.unload()   # libera memória (opcional)
```

```python
import lmstudio as lms

# Conecta no LM Studio
with lms.Client() as client:
    # Usa client.llm.model() para carregar o modelo diretamente
    model = client.llm.model(
        "oh-dcft-v3.1-gpt-4o-mini",   # nome/id do modelo no LM Studio
        config={
            "gpu": {
                "ratio": 0.7,          # metade GPU, metade CPU
            },
            "contextLength": 8192,    # contexto desejado
        }
    )

    # Gera a resposta
    response = model.respond("me da um codigo python de 40 linhas?")
    
    print("Resposta do modelo:")
    print(response.content)
    
    # Se quiser liberar memória depois
    # model.unload()
```
